```python
def pda(idx,inputStr,currentState,transitions,finalStates,stack):
    if idx == len(inputStr):
        if len(stack) == 0:
            return True
        if currentState in finalStates:
            return True

    if not stack:
        return False

    top = stack[-1]

    if idx<len(inputStr):
        key = (currentState,inputStr[idx],top)
        if key in transitions:
            for nextState,pushSym in transitions[key]:
                updStack = stack[:-1]
                if pushSym != "e":
                    for sym in reversed(pushSym):
                        updStack.append(sym)
                if pda(idx+1,inputStr,nextState,transitions,finalStates,updStack):
                    return True

    epskey = (currentState,"e",top)
    if epskey in transitions:
        for nextState,pushSym in transitions[epskey]:
            updStack = stack[:-1]
            if pushSym != "e":
                for sym in reversed(pushSym):
                    updStack.append(sym)
            if pda(idx,inputStr,nextState,transitions,finalStates,updStack):
                return True

    return False
```

```python
# initialising the stack with Z
stack = ["z"]
initialState = input("Enter the initial state: ")

temp = int(input("Enter the number of final states: "))
finalStates = []
for i in range(temp):
   finalStates.append(input(f"{i+1}-Enter the final state: ")) 

transitions = {}
temp = int(input("Enter the number of transitions: "))

for i in range(temp):
   st1 = input(f"{i+1}-Enter the current state: ")
   tsym = input(f"{i+1}-Enter the transition symbol: ")
   topStack = input(f"{i+1}-Enter the top of stack symbol: ")
   st2 = input(f"{i+1}-Enter the next state: ")
   pushSym = input(f"{i+1}-Enter the symbol to push onto stack: ")  # e is epsilon
   key=(st1,tsym,topStack)
   toAppend= (st2,pushSym)

   if key not in transitions: 
      transitions[key] = []  
   transitions[key].append(toAppend)



inputStr = input("Enter the input string: ")
if pda(0,inputStr,initialState,transitions,finalStates,stack):
    print(f"{inputStr} is accepted by the given PDA")
else:
    print(f"{inputStr} is rejected by the given PDA")
```

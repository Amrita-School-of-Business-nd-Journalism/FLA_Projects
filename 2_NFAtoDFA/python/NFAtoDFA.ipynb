```python
# change the dfatransistion - i to the value side

def nfatodfa(nfaStartstate,nfaTransistions,tsym):    
    temp = [set([nfaStartstate])]  
    visited = [set([nfaStartstate])]
    dfatransitions = {}

    while len(temp) > 0:
        currentState = temp.pop()
        for i in tsym:
            nextState = set()
            for state in currentState:
                    t1 = (state,i)
                    if t1 in nfaTransistions:
                        nextState = nextState.union(nfaTransistions[t1])
            dfatransitions[(frozenset(currentState),i)] = frozenset(nextState)  # set is mutable
            if nextState not in visited:
                    temp.append(nextState)
                    visited.append(nextState)
    return dfatransitions

def printDFA(nfaStartstate,dfaTransistions,nfaFinalStates):
    # Starting state 
    print(f"Starting State: {nfaStartstate}")
    print("\nDFA Transitions:")
    tempfinal = set()
    for (state, symbol), nextState in dfaTransistions.items():
        print(f"{state} --{symbol}--> {nextState}")
        if nfaFinalStates.intersection(state):
            tempfinal.add(state)
    print("\nDFA Final States:")
    for state in tempfinal:
        print(state)
```

```python
initial_state = input("Enter the initial state: ")

temp = int(input("Enter the number of final states: "))
final_state = set()
for i in range(temp):
    final_state.add(input(f"{i+1} - Enter the final state: "))

transitions = {}
tsym = set()
temp = int(input("Enter the number of transitions: "))

for i in range(temp):
    st = input(f"{i+1} - Enter state -1: ")
    st2 = input(f"{i+1} - Enter state -2: ")
    symbol = input(f"{i+1} - Enter the transition symbol: ")
    tsym.add(symbol)
    if (st, symbol) not in transitions:
        transitions[(st, symbol)] = []
    transitions[(st, symbol)].append(st2)

inputstr = input("Enter the input string: ")

dfatransitions = nfatodfa(initial_state,transitions,tsym)
printDFA(dfatransitions,initial_state,final_state)
```

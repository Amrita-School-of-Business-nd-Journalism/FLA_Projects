```python
def intersectionDFA(initial_state1,final_state1,transitions1,initial_state2, final_state2,transitions2,tsym):
    initial_state = (initial_state1,initial_state2)
    final_states = set()
    transitions = {} 

    queue = [initial_state]
    visited = {initial_state}

    while queue:
        current_state = queue.pop(0)

        state1,state2 = current_state

        if state1 in final_state1 and state2 in final_state2:
            final_states.add(current_state)

        for sym in tsym:
            next_state1 = transitions1.get((state1,sym))
            next_state2 = transitions2.get((state2,sym))

            if next_state1 is None or next_state2 is None:
                continue

            next_state = (next_state1,next_state2)

            transitions[(current_state,sym)] = next_state

            if next_state not in visited:
                visited.add(next_state)
                queue.append(next_state)

    return initial_state,transitions,final_states

def printDFA(initial_state,transitions,final_states):
    print("Initial State: ",initial_state)
    print("Final States: ",final_states)
    print("Transitions: ")
    for (state,sym),next_state in transitions.items():
        print(f"({state}---{sym}) -> {next_state}") 

```

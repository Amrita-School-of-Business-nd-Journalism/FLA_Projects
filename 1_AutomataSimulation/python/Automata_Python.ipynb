```python
def userdefineddfa(initial,transitions,final,inputstr):
    current_state = initial
    for i in inputstr:
        if (current_state,i) in transitions:
            current_state = transitions[(current_state,i)]
        else:
            return False

    return current_state in final


initial_state = input("Enter the initial state: ")
temp = int(input("Enter the number of final states: "))
final_state = []
for i in range(temp):
    final_state.append(input(f"{i} - Enter the final state: "))

transitions = {}
temp = int(input("Enter the number of transitions: "))
for i in range(temp):
    st = input(f"{i} - Enter state -1 ")
    st2 = input(f"{i} - Enter state -2 ")
    tsym = input(f"{i} - Enter the transition symbol: ")
    transitions[(st,tsym)] = st2

inputstr = input("Enter the input string: ")
if userdefineddfa(initial_state,transitions,final_state,inputstr):
    print(f"{inputstr} is accepted by the DFA")
else:
    print(f"{inputstr} is not accepted by the DFA")
```

```python
def nfa_userdefined(idx,inputstr,current,transitions,final):
    if len(inputstr) == idx:
        return current in final

    char = inputstr[idx]
    if (current,char) in transitions:
        for posstate in transitions[(current,char)]:
            if nfa_userdefined(idx+1,inputstr,posstate,transitions,final):
                return True
    return False

initial_state = input("Enter the initial state: ")
temp = int(input("Enter the number of final states: "))
final_state = []
for i in range(temp):
    final_state.append(input(f"{i+1} - Enter the final state: "))

transitions = {}
temp = int(input("Enter the number of transitions: "))
for i in range(temp):
    st = input(f"{i+1} - Enter state -1 ")
    st2 = input(f"{i+1} - Enter state -2 ")
    tsym = input(f"{i+1} - Enter the transition symbol: ")

    if (st,tsym) not in transitions:
        transitions[(st,tsym)] = []
    transitions[(st,tsym)].append(st2)

inputstr = input("Enter the input string: ")
if nfa_userdefined(0,inputstr,initial_state,transitions,final_state):
    print(f"{inputstr} is accepted by the NFA")
else:
    print(f"{inputstr} is not accepted by the NFA")
```

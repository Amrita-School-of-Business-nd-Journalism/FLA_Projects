```python
# Automata for checking numbers ending with 01

initial_state = 'q0'
final_state = 'q2'
intermediate_state = 'q1'

transitions = {
    ('q0','0'): 'q1',
    ('q0','1'): 'q0',
    ('q1','0'): 'q1',
    ('q1','1'): 'q2',
    ('q2','0'): 'q1',
    ('q2','1'): 'q0',
}


def dfa_checking(input_string):
    current_state = initial_state
    for i in input_string:
        if (current_state, i) in transitions:
            current_state = transitions[(current_state, i)]
        else:
            return False
    return current_state == final_state

print(dfa_checking("1101"))
print(dfa_checking("1010"))
```
